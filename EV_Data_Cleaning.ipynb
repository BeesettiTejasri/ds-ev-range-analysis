```python
import pandas as pd
df=pd.read_csv("/home/student/Downloads/ds-ev-range-analysis/data/electric_vehicles_spec_2025.csv")
print("Dataset Shape:")
print(df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nDataset Information:")
df.info()
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df["cargo_volume_l_original"]=df["cargo_volume_l"]
df["cargo_volume_l_clean"]=pd.to_numeric(df["cargo_volume_l"],errors="coerce")
print(df[["cargo_volume_l_original","cargo_volume_l_clean"]].head(20))
print(df["cargo_volume_l_clean"].isnull().sum())
```

```python

invalid_cargo = df[df["cargo_volume_l_clean"].isnull()& df["cargo_volume_l_original"].notnull()]
display(invalid_cargo[[
    "brand",
    "model",
    "cargo_volume_l_original"
]])
```

```python

positive_columns = [
    "top_speed_kmh",
    "battery_capacity_kWh",
    "efficiency_wh_per_km",
    "range_km",
    "acceleration_0_100_s",
    "seats",
    "length_mm",
    "width_mm",
    "height_mm"
]
for column in positive_columns:
    invalid = df[df[column] <= 0]
    print("\nColumn:", column)
    print("Invalid records:", len(invalid))
    if len(invalid) > 0:
        display(invalid[["brand", "model", column]])
```

```python
print(df["model"].isnull())
```

```python
display(df[df["model"].isnull()])
```

```python
df["model"]=df["model"].fillna("Unknown")
```

```python
display(df[df["model"].isnull()])
```

```python

text_columns = df.select_dtypes(include="object").columns
for column in text_columns:
    df[column] = df[column].str.strip()
print("Extra spaces cleaned from text columns.")
```

```python
print("Duplicate rows:",df.duplicated().sum())
```
